```python
from sklearn import tree # Paso 1: Se importa la librería trae desde sklearn, la cual sirve para utilizar el árbol de decisión.
from tabulate import tabulate # Paso 2: Se importa la librería tabulate, la cual sirve para tabular los datos, es decir ponerlos en una tabla de manera ordenada.
import pandas as pd # Paso 3: Se importa la librería pandas, la cual se utiliza para mostrar el resultado de una forma gráfica y ordenada en una tabla.
```

```python
# Crear el clasificador de árbol de decisiones
clasificador = tree.DecisionTreeClassifier() # Paso 4: Se define la variable clasificador, en la cual va a estar el árbol de decisión que se utiliza en esta actividad.
```

```python
# Datos de entrenamiento
# Paso 5: Se define una variable X, en donde se encuentra una lista compuesta de sub listas que a su vez cada sub lista almacena un conjunto de datos,
# en este caso cada conjunto tiene los síntomas de acuerdo a una enfermedad, por lo cual estos conjuntos van a ser comparados con las enfermedades propias
# de los síntomas en la siguiente variable.
X = [[1, 2, 3, 4, 5], [6, 7, 8, 4, 9], [10, 11, 1, 7, 12], [1, 3, 8, 13, 14], [4, 8, 9, 15, 16]]
# Paso 6: Definimos la variable Y, la cual contiene la lista con el nombre de cada una de las enfermedades que van a ser comparadas con los conjuntos de la variable X.
Y = ['Hipertensión arterial', 'Diabetes tipo 1', 'Hipoglucemia', 'Encefalitis', 'Esclerosis múltiple']
# Paso 7: Por medio del clasificador que fue definido previamente el cual corresponde al árbol de decisión, se procede a realizar la comparación de los conjuntos de la variable X,
# y las enfermedades de la variable Y.
clasificador = clasificador.fit(X, Y)
```

```python
# Solicitar nombre del usuario
# Paso 8: Se crea la variable Nombre a cuál va a almacenar el nombre del usuario, por lo cual se solicita este dato en forma de mensaje.
nombre = input("Por favor ingrese su nombre: ")
# Paso 9: Se muestra al usuario un mensaje con las instrucciones para seleccionar los síntomas que presenta, por medio de un mensaje.
print(f'Señor(a) {nombre}, de acuerdo con los siguientes síntomas presentados en orden numérico en la siguiente tabla, por favor digite el número de los síntomas que presente en este momento:\n')
```

```python
# Tabla de síntomas
# Paso 10: Se muestra al usuario los síntomas enumerados por medio de una tabla.
tabla_sintomas = [["1", "Dolor de cabeza fuerte"],
                  ["2", "Náuseas o vómitos"],
                  ["3", "Confusión"],
                  ["4", "Cambios en la visión o visión borrosa"],
                  ["5", "Sangrado nasal"],
                  ["6", "Aumento de la sed y de las ganas de orinar"],
                  ["7", "Aumento del apetito"],
                  ["8", "Fatiga"],
                  ["9", "Entumecimiento u hormigueo en las manos o los pies"],
                  ["10", "Palidez"],
                  ["11", "Temblores"],
                  ["12", "Latidos del corazón irregulares o acelerados"],
                  ["13", "Dolor en músculos o articulaciones"],
                  ["14", "Convulsiones"],
                  ["15", "Incapacidad para caminar"],
                  ["16", "Problemas cognitivos"]]
# Paso 11: Se imprime la tabla con sus encabezados, además del estilo para esta tabla.
print(tabulate(tabla_sintomas, headers=["ID", "Síntoma"], tablefmt="fancy_grid"))
print()
```

```python
# Solicitar síntomas al usuario
# Paso 12: Se le indica nuevamente al usuario por medio de un mensaje que digite los síntomas que padece.
sintomas = []
for i in range(5):
    sintoma = input(f"Digite el síntoma {i+1}: ")
    sintomas.append(sintoma)
print()
```

```python
# Mostrar tabla de síntomas ingresados
# Paso 13: Se crea una tabla en donde se presenta al usuario los síntomas seleccionados de manera enumerada.
tabla_sintomas_ingresados = []
for i, sintoma in enumerate(sintomas):
    tabla_sintomas_ingresados.append([str(i+1), sintoma])

# Paso 14: Se imprime la tabla junto con su encabezado y su estilo.
print(tabulate(tabla_sintomas_ingresados, headers=["Cantidad", "Síntomas que presenta el paciente"], tablefmt="pretty"))
print()
```

```python
# Realizar la predicción
# Paso 15: Se crea la variable predicción, en donde a partir de los síntomas que brindo el usuario, tomamos el árbol de decisión para predecir la enfermedad que tiene el usuario.
prediccion = clasificador.predict([sintomas])
```

```python
# Crear el DataFrame con el diagnóstico
# Paso 16: Creamos un diccionario de datos llamado tabladiagnostico con el nombre del usuario y la enfermedad diagnosticada.
tabladiagnostico = {'Nombre usuario': [nombre], 'Enfermedad diagnosticada': prediccion}
 # Paso 17: Creamos una variable llamada tabladiagnostico2 en donde mostramos la información del diccionario de datos tabladiagnostico en forma de tabla con ayuda de la librería Pandas.
tabladiagnostico2 = pd.DataFrame(tabladiagnostico)
# Paso 18: Imprimimos esta variable es decir la tabla con el resultado del diagnóstico.
print(tabladiagnostico2)
```
